```python
#Filter with relational operator
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
expensive=df[df['Price']>20000]
print(expensive)
```

```python
#Filter with relational operator(<)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
low_stock=df[df['Stock']<10]
print(low_stock)
```

```python
#Filter with relational operator(==)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
electronics=df[df['Category']=='Electronics']
print(electronics)
```

```python
#Filter with relational operator(!=)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
not_india=df[df['Country']!='India']
print(not_india)
```

```python
#Combine conditions with & (And)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
good_books=df[(df['Category']=='Books')&(df['Rating']>4.0)]
print(good_books)
```

```python
#Combine conditions with \(OR)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
toys_or_sports=df[(df['Category']=='Toys')|(df['Category']=='Sports')]
print(toys_or_sports)
```

```python
#Negate a condition with ~(NOT)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
out_of_stock=df[~(df['Stock']>0)]
print(out_of_stock)
```

```python
#Filtering with loc
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
result=df.loc[(df['Price']>=1000)&(df['Price']<=5000),['Product Name','Price']]
print(result)
```

```python
#Filtering with iloc(position_based)
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
first_20=df.iloc[0:20]
print(first_20[first_20['Category']=='Electronice'])
```

```python
#isin()-match against a list of values
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
target_countries=['USA','UK','Canada']
result=df[df['Country'].isin(target_countries)]
print(result)
```

```python
#~isin()-exclude  list of values
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
excluded=['Books','Beauty']
result=df[~df['Category'].isin(excluded)]
print(result)
```

```python
#unique()-list distinct values
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
print(df['Category'].unique())
```

```python
#unique()-count distinct values
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
print(df['Country'].value_counts())
```

```python
#value_counts()-frequesncy of each value
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
print(df['Category'].value_counts())
```

```python
#query()-filter within a string expression
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
result=df.query('Price>15000 and Stock>50')
print(result)
```

```python
#between()-filter within a numeric range
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
result=df[df['Rating'].between(4.0,5.0)]
print(result)
```

```python
#str.contains()-substring matching
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
result=df[df['Product Name'].str.contains('wireless',case=False)]
print(result)
```

```python
#duplicated()-find duplicate rows
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
check_cols=['Product Name','Category','Price','Rating','Stock','Country']
duplicate_rows=df[df.duplicated(subset=check_cols,keep='first')]
print(duplicate_rows)
```

```python
#nlnargest()/nsmallest()-top-N filltering
import pandas as pd
df=pd.read_excel('products_data.xlsx.xlsx')
top_5_expensive=df.nlargest(5,'Price')
bottom_5_rated=df.nsmallest(5,'Rating')
print(top_5_expensive)
print(bottom_5_rated)
```
