```python
#Inner join
import pandas as pd
d1={
    "Id":[1,2,3,4,5,6],
    "Name":["Pradhn","Venu","Madhurim","Nireekshan","Shafi","Veeru"],
    "Subject":["English","Java","Html","Python","C","dot net"]
}
df2={
    "Id":[11,12,13,14,15,16],
    "Name":["Srinu","Sumanth","Neelime","Daniel","Arjun","Veeru"],
    "Subject":["Java","Html","Cpp","Python","C","dot net"]
}
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
inn_join=pd.merge(df1,df2,on="Subject",how="inner")
print(df1)
print()
print(df2)
print()
print(inn_join)
```

```python
#left join
import pandas as pd
d1={
    "Id":[11,12,13,14,15,16],
    "Name":["Prashan","Venu","Madhurim","Nireekshan","Shafi","Veeru"],
    "Subject":["English","Java","Html","Python","C","dot net"]
}
d2={
    "Id":[11,12,13,14,15,16],
    "Name":["Srinu","Sumanth","Neelima","Daniel","Arjun","Veeru"],
    "Subject":["Java","Html","Cpp","Python","C","dot net"]
}
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
left_join=pd.merge(df1,df2,on="Subject",how="left")
print(df1)
print()
print(df2)
print()
print(left_join)
```

```python
#Right join
import pandas as pd
d1={
    "Id":[1,2,3,4,5,6],
    "Name":["Pradhan","Venu","Madhurima","Nireekshan","Shafi","Veeru"],
    "Subject":["English","Java","Html","Python","C","dot net"]
}
d2={
    "Id":[11,12,13,14,25,26],
    "Name":["Srinu","Sumanth","Neelima","Daniel","Arjun","Veeru"],
    "Subject":["Java","Html","Cpp","Python","C","dot net"]
}
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
right_join=pd.merge(df1,df2,on="Subject",how="right")
print(df1)
print()
print(df2)
print()
print(right_join)
```

```python
#Many to one
import pandas as pd
d1={
    "Employee":["Nireekshan","Veeru","Lavanya","Pradhan"],
    "Group":["Development","Testing","Testing","HR"]
}
d2={
    "Employee":["Lavanya","Nireekshan","Veeru","Pradhan"],
    "Hire_data":[2010,2012,2014,2016]
}
d3={
    "Group":["Testing","Development","HR"],
    "Supervisor":["Shafi","Daniel","Neelima"]
}
df1=pd.DataFrame(d1)
df2=pd.DataFrame(d2)
df3=pd.DataFrame(d3)
one_one=pd.merge(df1,df2)
many_one=pd.merge(one_one,df3)
print(df1)
print()
print(df2)
print()
print(many_one)
```
