```python
#Creating a DataFrame
import pandas as pd
import numpy as np
data=[
    ['Shahid',21,40000],
    ['Priya',23,50000],
    ['Balu',27,60000],
    ['Veeru',45,90000],
    ['np.nan',2,99000],
    ['Prasad',1,41000]
]
c=['Name','Age','Salary']
df1=pd.DataFrame(data,columns=c)
print(df1)
```

```python
#notnull()method
import pandas as pd
import numpy as np
data=[
    ['Priya',21,40000],
    ['Balaji',22,20000],
    ['Veeru',45,90000],
    ['Sumanth',20,95000],
    [np.nan,2,99000],
    ['Prasad',1,41000]
]
c=['Name','Age','Salary']
df1=pd.DataFrame(data,columns=c)
d=df1.Name.notnull()
df2=df1[d]
print(df1)
print()
print(df2)
```
