```python
#Explore column names
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
print(df.columns)
```

```python
#columns attribute->list
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
col_list=list(df.columns)
print(col_list)
```

```python
#rename() with a dictinory
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df_renamed=df.rename(columns={'Student ID':'StudentId','Student Name':'StudentName'})
print(df.columns)
print(df_renamed.columns)
```

```python
#rename() wih inplace=True
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.rename(columns={'Marks':'Score'},inplace=True)
print(df.columns)
```

```python
#rename() using axis parameter
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.rename({'Grade':'FinalGrade'},axis=1,inplace=True)
print(df.columns)
```

```python
#Convert all column names to UPPERCASE
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.columns=df.columns.str.upper()
print(df.columns)
```

```python
#Convert ll column names to lowercase
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.columns=df.columns.str.lower()
print(df.columns)
```

```python
#Strip extra spaces from column names
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.columns=df.columns.str.strip()
print(df.columns)
```

```python
#Replace spaces with undersores
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.columns=df.columns.str.replace('','_')
print(df.columns)
```

```python
#Assign s brand_new column list directly
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df.columns=['id','name','age','city','subject','score','grade']
print(df.columns)
```

```python
#add_perfix()
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df=df.add_prefix('col_')
print(df.columns)
```

```python
#add_suffix()
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
df=df.add_suffix('_2024')
print(df.columns)
```

```python
#inplace=False VS inplace=True
import pandas as pd
df=pd.read_excel('student_data.xlsx.xlsx')
result=df.rename(columns={'age':'Agg'})
print(df.columns)
print(result.columns)
output=df.rename(columns={'age':'Age'},inplace=True)
print(output)
print(df.columns)
```
