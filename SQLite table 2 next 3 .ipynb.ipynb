```python
#Employee Audit Table Cleanup
import sqlite3
conn=sqlite3.connect(':memory:')
cur=conn.cursor()
cur.execute("""
CREATE TABLE employees(
emp_id INTEGER PRIMARY KEY,
name TEXT,
salary INTEGER
)
""")
cur.execute("ALTER TABLE employees ADD COLUMN department TEXT")
cur.executemany("INSERT INTO employees VALUES(?,?,?,?)",[
    (1,'Asha',50000,'IT'),
    (2,'Bala',-2000,'HR'),
    (3,'Chitra','None','IT'),
    (4,'Deepak',60000,'Finance')
])
cur.execute("DELETE FROM employees WHERE salary IS NULL OR salary < 0")
for row in cur.execute("SELECT * FROM employees"):
  print(row)
```

```python
#Top Paid Employees by Department
import sqlite3
con=sqlite3.connect(":memory:")
cur=con.cursor()
cur.execute("""
CREATE TABLE IF NOT EXISTS staff(
emp_name TEXT,
department TEXT,
salary INTEGER
)
""")
cur.executemany("INSERT INTO staff VALUES(?,?,?)",[
    ('Ravi','IT',70000),
    ('Meena','IT',70000),
    ('Suresh','IT',50000),
    ('Anita','HR',60000),
    ('Karthik','HR',55000)
])
query="""
SELECT emp_name,department,salary,rank FROM(
SELECT emp_name,department,salary,
DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS rank
FROM staff
)
WHERE rank <=2
"""
for row in cur.execute(query):
  print(row)
print()
#Deletect Duplicate Customer Records
cur.execute("""
CREATE TABLE customers(
customers_name TEXT,
order_id INTEGER
)
""")
cur.executemany("INSERT INTO customers VALUES(?,?)",[
    ('Remesh',101),
    ('Sita',102),
    ('Ramesh',103),
    ('Sita',104),
    ('Ajay',105)
])
query="""
SELECT customers_name AS Name,COUNT(*) AS Total_Orders
FROM customers
GROUP BY customers_name
HAVING COUNT(*) >1
"""
for row in cur.execute(query):
  print(row)
print()
cur.execute("""
UPDATE staff
SET salary=CASE
WHEN department='IT' THEN salary *1.10
WHEN department='HR' THEN salary * 1.05
ELSE salary
END
""")
con.commit()
for row in cur.execute("SELECT emp_name,department,salary FROM staff"):
  print(row)
print()
query="""
SELECT emp_name,salary
FROM staff
WHERE department ='HR'
"""
for row in cur.execute(query):
  print(row)
```
