```python
class BankAccount:
  def __init__(self,owner,balance=0):
    self.owner=owner
    self.__balance=balance
  def deposit(self,amount):
    if amount<=0:
      print("Deposit mount must be positive.")
      return
    self.__balance+=amount
    print(f"Deposited {amount}.New balance:{self.__balance}")
  def withdraw(self,amount):
    if amount>self.__balance:
      print("Insufficiant funds.")
      return
    self.__balance-=amount
    print(f"withdrew {amount}.New balance:{self.__balance}")
  def get_balance(self):
    return self.__balance
account=BankAccount("Ravi",1000)
account.deposit(500)
account.withdraw(2000)
account.withdraw(300)
print("Final balance:",account.get_balance())
```

```python
class Employee:
  def __init__(self,name,salary):
    self.name=name
    self.__salary=salary
  @property
  def salary(self):
    return self.__salary
  @salary.setter
  def salary(self,value):
    if value<0:
      raise ValueError("Salary cannot be negative.")
      self.__salary=value
emp=Employee("Anita",50000)
print("Initial salary:",emp.salary)
emp.salary=60000
print("Updated salary:",emp.salary)
try:
  emp.salary=-5000
except ValueError as e:
  print("Error:",e)

```

```python
class Shape:
  def area(self):
    raise NotlmplementedError("Subclasses must implement are()")
class Circle(Shape):
  def __init__(self,radius):
    self.radius=radius
  def area(self):
    return 3.14159*self.radius**2
class Rectangle(Shape):
  def __init__(self,length,width):
    self.length=length
    self.width=width
  def area(self):
    return self.length*self.width
class Triangle(Shape):
  def __init__(self,base,height):
    self.base=base
    self.height=height
  def area(self):
    return 0.5*self.base*self.height
shapes=[Circle(5),Rectangle(4,6),Triangle(3,8)]
for shape in shapes:
  print(f"{shape.__class__.__name__}area:{shape.area():.2f}")
```

```python
class Student:
  def __init__(self,name,marks):
    self.name=name
    self.__marks=marks
  @property
  def marks(self):
    return self.__marks
  @marks.setter
  def marks(self,value):
    if not(0<=value<=100):
       raise ValueError("Marks must be between 0 and 100.")
    self.__marks=value
  @property
  def grade(self):
    if self.__marks>=90:
      return "A"
    elif self.__marks>=75:
      return"B"
    elif self.__marks>=50:
      return "C"
    return"F"
s1=Student("Priya",88)
print(f"{s1.name} scored{s1.marks}, Grade:{s1.grade}")
s1.marks=95
print(f"{s1.name} scored{s1.marks}, Grade:{s1.grade}")


```

```python
class Vector:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __add__(self,other):
    return Vector(self.x + other.x,self.y+ other.y)
  def __str__(self):
    return f"Vector({self.x},{self.y})"
v1=Vector(2,3)
v2=Vector(4,1)
v3=v1+v2
print("v1=",v1)
print("v2=",v2)
print("v1+v2=",v3)
```

```python
class Book:
  def __init__(self,title,copies):
    self.title=title
    self.__copies=copies
  def borrow(self):
    if self.__copies<=0:
      print(f"{self.title}'is currently unavailable.")
      return
    self.__copies-=1
    print(f"You borrowed'{self.title}'.Copies left:{self.__copies}")
  def return_book(self):
    self.__copies+=1
    print(f"'{self.title}'returned.copies left:{self.__copies}")
  def available_copies(self):
    return self.__copies
book=Book("Python Basics",2)
book.borrow()
book.borrow()
book.borrow()
book.return_book()
print("Available copies:",book.available_copies())



```

```python
from abc import ABC,abstractmethod
class Payment(ABC):
  @abstractmethod
  def pay(self,amount):
    pass
class CreditCardPayment(Payment):
  def pay(self,amount):
    print(f"Paid{amount}using Credit Card.")
class UpiPayment(Payment):
  def pay(self,amount):
    print(f"paid{amount}using UPI.")
def process_payment(payment_method,amount):
  payment_method.pay(amount)
payments=[CreditCardPayment(),UpiPayment()]
for method in payments:
  process_payment(method, 1500)
```

```python
class Temperature:
  def __init__(self,celsius=0):
    self.__celsius=celsius
  @property
  def celsius(self):
    return self.__celsius
  @celsius.setter
  def celsius(self,value):
    if value<-273.15:
      raise ValueError("Temperature below absolute zero is not possible.")
      self.__celsius=value
  @property
  def fahrenheit(self):
    return(self.__celsius*9/5)+32
  @fahrenheit.setter
  def fahrenheit(self,value):
    self.celsius=(value-32)*5/9
t=Temperature(25)
print(f"Celsius:{t.celsius},Fahrenheit:{t.fahrenheit}")
t.fahrenheit=98.6
print(f"Celsius:{t.celsius:.1f},Fahrenheit:{t.fahrenheit}")
```

```python
class DiscountCalculator:
  def calculate(self,price,discount_percent=0,flat_discount=0):
    amount=price-(price*discount_percent/100)-flat_discount
    return max(amount,0)
calc=DiscountCalculator()
print("Price only:",calc.calculate(1000))
print("with percent discount:",calc.calculate(1000,discount_percent=10))
print("with flat discount:",calc.calculate(1000,flat_discount=150))
print("with both discounts:",calc.calculate(1000,discount_percent=10,flat_discount=50))

```

```python
class Inventory:
  def __init__(self,item_name,stock=0):
    self.item_name=item_name
    self.__stock=stock
  def add_stock(self,quantity):
    if quantity<=0:
      print("Quantity to add must be positive.")
      return
    self.__stock+=quantity
    print(f"Added {quantity}.stock:{self.__stock}")
  def remove_stock(self,quantity):
    if quantity>self.__stock:
      print("Not enough stock to remove.")
      return
    self.__stock-=quantity
    print(f"Removed {quantity}.Stock:{self.__stock}")
  def check_stock(self):
    return self.__stock
inv=Inventory("Laptop",10)
inv.add_stock(5)
inv.remove_stock(20)
inv.remove_stock(8)
print("Final stock:",inv.check_stock())

```

```python
class Employee:
  def __init__(self,name,base_salary):
    self.name=name
    self.base_salary=base_salary
  def calculate_salary(self):
    return self.base_salary
class Manager(Employee):
  def calcuate_salary(self):
    return self.base_salary+10000
class Developer(Employee):
  def __init__(self,name,base_salary,projects_completed):
    super().__init__(name,base_salary)
    self.projects_completed=projects_completed
  def calculate_salary(self):
    return self.base_salary+(self.projects_completed*2000)
employees=[Manager("Balaji",60000),Developer("Priyanga",50000,3),Employee("Suresh",40000),]
for emp in employees:
  print(f"{emp.name}({emp.__class__.__name__}):{emp.calculate_salary()}")
```

```python
class Vehicle:
  daily_rate=0
  def __init__(self,name):
    self.name=name
  def rental_cost(self,days):
    return self.daily_rate*days
class Car(Vehicle):
  daily_rate=1500
class Bike(Vehicle):
  daily_rate=400
class Truck(Vehicle):
  daily_rate=3000
fleet=[Car("Sedan"),Bike("Sports bike"),Truck("Mini Truck")]
rental_days=3
total=0
for vehicle in fleet:
  cost=vehicle.rental_cost(rental_days)
  total+=cost
  print(f"{vehicle.name}({vehicle.__class__.__name__}):{cost}")
print("Total rental cost:",total)


```

```python
class Account:
  def __init__(self,owner,balance):
    self.owner=owner
    self._balance=balance
  def apply_monthy_adjustment(self):
    raise NotlmplementedError("Subclasses must implement this method")
  def get_balance(self):
    return self._balance
class SavingsAccount(Account):
  interest_rate=0.04
  def apply_monthly_adjustment(self):
    interest=self._balance*self.interest_rate/12
    self._balance+=interest
    print(f"{self.owner}:interest added,new balance:{self._balance:.2f}")
class CurrentAccount(Account):
  maintenance_fee=50
  def apply_monthly_adjustment(self):
    self._balance-=self.maintenance_fee
    print(f"{self.owner}:fee deducted,new balance:{self._balance:.2f}")
accounts=[SavingsAccount("Meera",100000),CurrentAccount("Vikram",25000),]
for acc in accounts:
  acc.apply_monthly_adjustment()
  print(f"Balance via get_balnce():{acc.get_balance():.2f}")
```
