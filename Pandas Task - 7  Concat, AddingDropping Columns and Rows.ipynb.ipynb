```python
print(1)
import pandas as pd
df = pd.read_excel('books_data.xlsx')
fantasy = df[df['Genre'] == 'Fantasy']
thriller = df[df['Genre'] == 'Thriller']
combined = pd.concat([fantasy, thriller])
print(combined.shape)
print(2)
combined = pd.concat([fantasy, thriller], ignore_index=True)
print(combined.index)
print(3)
combined = pd.concat([fantasy, thriller], keys=['Fantasy', 'Thriller'])
print(combined.index.get_level_values(0).unique())
print(4)
top5 = df.head(5)
discount = pd.DataFrame({'Discount %': [10, 15, 5, 20, 0]}, index=top5.index)
with_discount = pd.concat([top5, discount], axis=1)
print(with_discount)
print(5)
df['Price After Tax'] = df['Price'] * 1.05
print(df[['Price', 'Price After Tax']].head())
print(6)
df.insert(7, 'Is Recent', df['Publish Year'] >= 2015)
print(df.columns.tolist())
print(7)
df_with_flag = df.assign(High_Rated=df['Rating'] >= 4.5)
print('High_Rated' in df.columns)
print('High_Rated' in df_with_flag.columns)
print(8)
df = df.assign(
    Price_Rounded=df['Price'].round(0),
    Decade=(df['Publish Year'] // 10) * 10,
)
print(df[['Price', 'Price_Rounded', 'Publish Year', 'Decade']].head())
print(9)
df = df.drop(columns=['Is Recent'])
print(df.columns.tolist())
print(10)
df = df.drop('Decade', axis=1)
print(df.columns.tolist())
print(11)
del df['Price_Rounded']
print(df.columns.tolist())
print(12)
tax_prices = df.pop('Price After Tax')
print(tax_prices.head())
print('Price After Tax' in df.columns)
print(13)
df['Temp Flag'] = 'x'
df.drop(columns=['Temp Flag'], inplace=True)
print('Temp Flag' in df.columns)
print(14)
df_fewer_rows = df.drop([0, 1, 2])
print(df_fewer_rows.shape)
print(15)
low_rated_index = df[df['Rating'] < 3.0].index
df_cleaned = df.drop(low_rated_index)
print(df_cleaned.shape)
print(16)
check_cols = df.columns.drop('Book ID')
print('Before:', df.shape)
df_unique = df.drop_duplicates(subset=check_cols)
print('After:', df_unique.shape)
print(17)
df_unique_titles = df.drop_duplicates(subset=['Title', 'Author'])
print(df_unique_titles.shape)
print(18)
df_cleaned = df_cleaned.reset_index(drop=True)
print(df_cleaned.index)
print(19)
check_cols = df.columns.drop('Book ID')
first_genres = ['Mystery', 'Romance', 'Science Fiction', 'Fantasy']
part1 = df[df['Genre'].isin(first_genres)].drop_duplicates(subset=check_cols)
part2 = df[~df['Genre'].isin(first_genres)].drop_duplicates(subset=check_cols)
rebuilt = pd.concat([part1, part2], ignore_index=True)
print(rebuilt.shape)
```
