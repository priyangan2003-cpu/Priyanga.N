```python
#Delect missing values with isna()
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
print(df.isna())
```

```python
#Count missing values per column
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
print(df.isna().sum())
```

```python
#notnull() to find vaild rows
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
print(df.notnull())
```

```python
#Filter rows with a missing value in a specific column
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
missing_salary=df[df['Salary'].isna()]
print(missing_salary)
```

```python
#Fillter rows with complete data in a spexific column
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
valid_dept=df[df['Department'].notnull()]
print(valid_dept)
```

```python
#Check if ANY missing values exist
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
print(df.isna().values.any())
```

```python
#dropna()-drop rows with any missing value
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df_clean=df.dropna()
print(df_clean.shape)
```

```python
#dropna(how='all')
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df_no_empty_rows=df.dropna(how='all')
print(df_no_empty_rows.shape)
```

```python
#dropna(subset=[...])
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df_key_cols=df.dropna(subset=['Salary','Age'])
print(df_key_cols.shape)
```

```python
#dropna(axis=1)
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df_full_columns=df.dropna(axis=1)
print(df_full_columns.columns)
```

```python
#dropna(theresh=...)
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df_thresh=df.dropna(thresh=5)
print(df_thresh.shape)
```

```python
#fillna() with a constant value
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df['Department']=df['Department'].fillna('Unknow')
print(df['Department'].isna().sum())
```

```python
#fillna() with column mean/median
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df['Salary']=df['Salary'].fillna(df['Salary'].mean())
df['Age']=df['Age'].fillna(df['Age'].median())
print(df[['Salary','Age']].isna().sum())
```

```python
#fillna(method='ffill') and bfill()
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df['Join Date']=df['Join Date'].fillna(method='ffill')
print(df['Join Date'].isna().sum())
```

```python
#fillna() with a dictionary
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df=df.fillna({'Performance Score':3.0,'Employee Name':'Not Provied'})
print(df.isna().sum())
```

```python
#interpolate
import pandas as pd
df=pd.read_excel('employee_data.xlsx')
df['Performance Score']=df['Performance Score'].interpolate()
print(df['Performance Score'].isna().sum())
```
