```python
#E-Commerce Order Deduplication and Priority Processing
orders=["ORD1404","ORD101","ORD104","ORD108","ORD101","ORD108"]
seen=set()
unique_orders=[]
duplicates=set()
for order_id in orders:
  if order_id in seen:
    duplicates.add(order_id)
  else:
    seen.add(order_id)
    unique_orders.append(order_id)
unique_order_set=set(unique_orders)
print("Processing Queue:",unique_orders)
print("Unique Orders:",unique_order_set)
print("Duplicates:",duplicates)

```

```python
#Employee Skill Matching Using sets and Ductionaries
employee_skills={
    "Anu":{"Python","SQL","Power BI","Excel"},
    "Bala":{"Python","SQL"},
    "Charan":{"Python","SQL","Spark","AWS"},
    "Divya":{"Excel","Power BI","SQL"}
}
required={"Python","SQL","Power BI"}
fully_qualified=[]
partial_candidates={}
for employee,skills in employee_skills.items():
  missing=required-skills
  matched=required&skills
  if not missing:
     fully_qualified.append(employee)
  if len(matched)>=2:
     partial_candidates[employee]=matched
  print(employee,"Missing:",missing)
  print("Fully Qualified:",fully_qualified)
  print("Partil Candidates:",partial_candidates)

```

```python
#Immutable Transaction Records and safe dictionary keys
transactions=[(101,"SAVINGS","CREDIT",5000),(102,"CURRENT","DEBIT",2500),(101,"SAVINGS","DEBIT",1000),(101,"SAVINGS","CREDIT",5000)]
unique_transactions=set(transactions)
transaction_map={}
duplicate_records=set()
for record in transactions:
  key=record[:3]
  if key in transaction_map:
    duplicate_records.add(record)
    transaction_map[key]=record[3]
print("Unique Transactions:",unique_transactions)
print("Transaction Map:",transaction_map)
print("Repeated Records:",duplicate_records)
```

```python
#Inventory Aggregation with Nested Dictionaries
movements=[("Coimbatore","Laptop",5),("Chennai","Laptop",3),("Coimbatore","Mouse",10),("Chennai","laptop",-1),("coimbatore","Loptop",-2),("Cjhennai","Mouse",4)]
inventory={}
for store,product,qty in movements:
  inventory.setdefault(store,{})
  inventory[store][product]=inventory[store].get(product,0)+qty
print("Inventory:",inventory)
alerts=[]
for store,products in inventory.items():
  for product,qty in products.items():
    if qty<=0:
      print("OUT OF STOCK:",store,product)
    if qty<3:
       alerts.append((store,product,qty))
print("Low Stock Alearts:",alerts)



```

```python
#Log Analysis:Counting Events,Ranking,and Unique Users
logs=[
  ("u101","LOGIN"),
  ("u102","LOGIN"),
  ("u101","PURCHASE"),
  ("u103","LOGIN"),
  ("u101","PURCHASE"),
  ("u102","LOGOUT"),
  ("u103","PURCHASE"),
  ("u101","LOGIN"),
  ("u103","PURCHASE")
]
event_counts={}
unique_users=set()
user_events={}
purchase_counts={}
for user,event in logs:
  unique_users.add(user)
  event_counts[event]=event_counts.get(event,0)+1
  user_events.setdefault(user,set())
  user_events[user].add(event)
  if event=="PURCHASE":
    purchase_counts[user]=purchase_counts.get(user,0)+1
max_purchase=max(purchase_counts.values())
top_users=[
    user for user,count in purchase_counts.items()
    if count++max_purchase
]
result={
    "event_counts":event_counts,
    "unique_users":unique_users,
    "user_events":user_events,
    "top_purchase_users":top_users
}
print(result)
```
