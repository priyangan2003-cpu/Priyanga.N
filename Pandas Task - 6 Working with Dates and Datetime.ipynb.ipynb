```python
import pandas as pd
df = pd.read_excel('transactions_data.xlsx')
print(df.dtypes)
print(df.info())
print()
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'], errors='coerce')
print(df['Transaction Date'].dtype)
print()
raw = pd.read_excel('transactions_data.xlsx', sheet_name='Transactions_Data')
try:
      raw['Transaction Date'] = raw['Transaction Date'].astype('datetime64[ns]')
except Exception as e:
      print('astype() failed on messy data:', type(e).__name__)
print()
print(df['Transaction Date'].isna().sum())
missing_dates = df[df['Transaction Date'].isna()]
print(missing_dates[['Transaction ID','Transaction Date']])
print()
df_valid = df.dropna(subset=['Transaction Date'])
print(df_valid.shape)
print()
df_valid['Year'] = df_valid['Transaction Date'].dt.year
print(df_valid[['Transaction Date','Year']].head())
print()
df_valid['Month'] = df_valid['Transaction Date'].dt.month
print(df_valid[['Transaction Date','Month']].head())
print()
df_valid['Day'] = df_valid['Transaction Date'].dt.day
print(df_valid[['Transaction Date','Day']].head())
print()
df_valid['Month Name'] = df_valid['Transaction Date'].dt.month_name()
print(df_valid[['Transaction Date','Month Name']].head())
print()
df_valid['Day of Week'] = df_valid['Transaction Date'].dt.day_name()
print(df_valid[['Transaction Date','Day of Week']].head())
print()
df_valid['Weekday Num'] = df_valid['Transaction Date'].dt.dayofweek
print(df_valid[['Transaction Date','Day of Week','Weekday Num']].head())
print()
df_valid['Is Weekend'] = df_valid['Weekday Num'].isin([5, 6])
print(df_valid[['Transaction Date','Day of Week','Is Weekend']].head(10))
print()
df_valid['Quarter'] = df_valid['Transaction Date'].dt.quarter
print(df_valid[['Transaction Date','Quarter']].head())
print()
latest_date = df_valid['Transaction Date'].max()
cutoff_20d = latest_date - pd.Timedelta(days=20)
last_20_days = df_valid[df_valid['Transaction Date'] >= cutoff_20d]
print('Latest date:', latest_date)
print('Cutoff date:', cutoff_20d)
print(last_20_days.shape)
print()
latest_date = df_valid['Transaction Date'].max()
cutoff_2m = latest_date - pd.DateOffset(months=2)
last_2_months = df_valid[df_valid['Transaction Date'] >= cutoff_2m]
print(last_2_months.shape)
print()
latest_date = df_valid['Transaction Date'].max()
cutoff_2y = latest_date - pd.DateOffset(years=2)
last_2_years = df_valid[df_valid['Transaction Date'] >= cutoff_2y]
print(last_2_years.shape)
print()
summer_2023 = df_valid[df_valid['Transaction Date'].between(
        '2023-06-01', '2023-08-31')]
print(summer_2023.shape)
print()
sorted_recent_first = df_valid.sort_values('Transaction Date', ascending=False)
print(sorted_recent_first[['Transaction ID','Transaction Date']].head())
print()
monthly_counts = df_valid.groupby(
        df_valid['Transaction Date'].dt.to_period('M')).size()
print(monthly_counts)
```
