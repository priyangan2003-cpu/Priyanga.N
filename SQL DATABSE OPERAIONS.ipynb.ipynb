```python
#Create  datbasecalled student_db
import sqlite3
import os
#step 1: Create Database
conn=sqlite3.connect("student_db.sqlite")
cursor=conn.cursor()
#step 2: Create table
cursor.execute("DROP TABLE IF EXISTS students")
cursor.execute("""CREATE TABLE Students(
id INTEGER PRIMARY KEY,
name TEXT,
age INTEGER,
marks INTEGER
)
""")
#step 3: insert Records
cursor.execute("INSERT INTO students VALUES(1,'Priya',23,85)")
cursor.execute("INSERT INTO students VALUES(2,'Balaji',19,90)")
cursor.execute("INSERT INTO students VALUES(3,'Aadhira',18,78)")
conn.commit()
#step 4: Update Marks
cursor.execute("UPDATE students SET marks=88 WHERE name='Balaji'")
conn.commit()
#step 5: Delete one Row
cursor.execute("DELETE FROM students WHERE id=1")
conn.commit()
#step 6: Display output
cursor.execute("SELECT*FROM students")
rows=cursor.fetchall()
print("Final Table Output:")
for row in rows:
  print(row)
conn.close()
#step 7: Rename Database
os.rename("student_db.sqlite","student_database_renamed.sqlite")
#sete 8: Delete Database
os.remove("student_database_renamed.sqlite")
```

```python
#Crete a database employee_db with a table wmployees having:emp_id,emp_name,deprtment,salary.
import sqlite3
import os
conn=sqlite3.connect("employee_db.sqlite")
cursor=conn.cursor()
cursor.execute("DROP TABLE IF EXISTS employees")
cursor.execute("""
CREATE TABLE employees(
emp_id INTEGER PRIMARY KEY,
emp_name TEXT,
department TEXT,
salary INTEGER
)
""")
cursor.executemany(
    "INSERT INTO employees VALUES(?,?,?,?)",
    [
        (1,"Priya","IT",40000),
        (2,"Balaji","HR",28000),
        (3,"Saravana","IT",35000),
        (4,"Ravi","Finance",30000)
    ]
)
conn.commit()
#Update Salary for IT department
cursor.execute("""
UPDATE employees
SET salary=salary*1.10
WHERE department='IT'
""")
conn.commit()
#Delete employees with salary<30000
cursor.execute("DELETE FROM employees WHERE salary < 30000")
conn.commit()
cursor.execute("SELECT*FROM employees")
print("Final Output:")
for row in cursor.fetchall():
  print(row)
conn.close()
os.rename("employee_db.sqlite","employee_db_final.sqlite")
os.remove("employee_db_final.sqlite")
```

```python
import sqlite3
import os
conn=sqlite3.connect("product_db.sqlite")
cursor=conn.cursor()
cursor.execute("DROP TABLE IF EXISTS products")
cursor.execute("""
CREATE TABLE products(
product_id INTEGER PRIMRY KRY,
product_name TEXT,
price INTEGER,
stock INTEGER
)
""")
cursor.executemany(
    "INSERT INTO products VALUES(?,?,?,?)",
    [
        (1,"Laptop",50000,10),
        (2,"Mouse",500,0),
        (3,"Keyboard",1500,25),
        (4,"Monitor",12000,15),
        (5,"Printer",8000,5)
    ]
)
conn.commit()
#Update price for low stock
cursor.execute("""
UPDATE products
SET price=price+500
WHERE stock <20
""")
conn.commit()
#Delete out_of_stock products
cursor.execute("DELETE FROM products WHERE stock=0")
conn.commit()
cursor.execute("SELECT*FROM products")
print("Final Product Table:")
for row in cursor.fetchall():
  print(row)
conn.close()
os.rename("product_db.sqlite","product_db_backup.sqlite")
os.remove("product_db_backup.sqlite")
```
