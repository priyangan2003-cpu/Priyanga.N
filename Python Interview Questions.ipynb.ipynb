```python
#Reverse  String
def reverse_string(s):
  result=""
  for ch in s:
    result=ch+result
  return result
def reverse_string_slicing(s):
  return s[::-1]
print(reverse_string('hello'))
print(reverse_string_slicing('hello'))
```

```python
#Check if a Number is Prime
def is_prime(n):
  if n <2:
    return False
  for i in range(2,int(n**0.5)+1):
    if n% i==0:
      return False
  return True
num=30
num1=7
print('Prime' if is_prime(num) else 'Not Prime')
print('Prime' if is_prime(num1) else 'Not Prime')
```

```python
#Generate the Fibnacci Series
def fibonacci(n):
  series=[]
  a,b = 0,1
  for _ in range(n):
    series.append(a)
    a,b=b, a+b
  return series
print(fibonacci(5))
```

```python
#Factorial of Number
def factorial(n):
  if n==0 or n==1:
    return 1
  return n * factorial(n-1)
print(factorial(3))
```

```python
#Check Plindrome String
def is_palindrome(s):
  s =s.lower().replace('',"")
  return s==s[::-1]
print(is_palindrome('madam'))
print(is_palindrome('dad'))
print(is_palindrome('A man a plan a canal Pnama'))
```

```python
#Find the Largest of Three Numbers
def largest_of_three(a,b,c):
  return max(a,b,c)
def lrgest_manual(a,b,c):
  largest=a
  if b > largest:
    largest=b
  if c > largest:
    largest=c
  return largest
print(largest_of_three(12,45,30))
```

```python
#Count Vowels and Consonats in a String
def count_vowels_consonants(s):
  vowels='aeiouAEIOU'
  v_count=c_count=0
  for ch in s:
    if ch.isalpha():
      if ch in vowels:
        v_count +=1
      else:
        c_count +=1
  return v_count,c_count
v,c=count_vowels_consonants('interview')
print(f'Vowels={v}, Consonants={c}')
```

```python
#Check Armstrong Number
def is_armstrong(n):
  digits=str(n)
  power=len(digits)
  total=sum(int(d)**power for d in digits)
  return total==n
num=153
print('Armstrong Number' if is_armstrong(num) else 'Not an Armstrong Number')
```

```python
#Remove Duplicats from a list
def remove_duplicates(lst):
  seen=set()
  result=[]
  for item in lst:
    if item not in seen:
      seen.add(item)
      result.append(item)
  return result
print(remove_duplicates([1,2,2,3,4,4,5]))
```

```python
#Find the Second largest Numbers in  List
def second_largest(ist):
  unique = list(set(ist))
  unique.sort(reverse=True)
  if len(unique) <2:
    return None
  return unique[1]
print(second_largest([10,25,3,47,47,8]))
```

```python
#Check if two Strings are Anagrams
def is_anagram(s1,s2):
  s1=s1.replace('',"").lower()
  s2=s2.replace('',"").lower()
  return sorted(s1)==sorted(s2)
print(is_anagram('listen','silent'))
print(is_anagram('hellow','world'))
```

```python
#FizzBuzz
def fizzbuzz(n):
  result=[]
  for i in range(1,n+1):
    if i % 15==0:
      result.append('FizzBuzz')
    elif i % 3==0:
      result.append('Fizz')
    elif i % 5==0:
      result.append('Buzz')
    else:
      result.append(str(i))
  return result
print(''.join(fizzbuzz(15)))
```

```python
#Find the GCD(HCF) of Two Numbers
def gcd(a,b):
  while b:
    a,b = b,a % b
  return a
print(gcd(48,18))
import math
print(math.gcd(48,18))
```

```python
#Count the Frequency of Each Element in a List
def count_frequency(lst):
  freq={}
  for item in lst:
    freq[item]=freq.get(item,0)+1
  return freq
print(count_frequency(['a','b','a','c','b','a']))
from collections import Counter
print(Counter(['a','b','a','c','b','a']))

```
