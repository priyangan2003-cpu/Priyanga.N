```python
#Student Management System
students=[]
def get_grade(average):
  if average >=90:return'A+'
  elif average >=80:return'A'
  elif average >=70:return'B'
  elif average >=60:return'C'
  elif average >=50:return'D'
  else:return'F'
def add_student():
    roll=int(input('Enter Roll Number:'))
    name=input('Enter Name:').strip()
    age=int(input('Enter Age:'))
    marks=[]
    subject=['Tamil','English','Maths','Science','Social']
    for sub in subject:
      while True:
        try:
           m=float(input(f'{sub} marks(0-100):'))
           if 0<=m<=100:
               marks.append(m)
               break
           else:
             print('Marks must be between 0 and 100.')
        except ValueError:
          print('Please enter a number.')
    student={'roll':roll,'name':name,'age':age,'marks':marks}
    students.append(student)
    print(f'Student {name} added successfully!')
def view_all_students():
  if not students:
    print('No student records found.')
    return
  print(f"{'Roll':<6}{'Name':<18}{'Age':<5}{'Total':<8}{'Avg':<8}{'Grade'}")
  print('-'*55)
  for s in students:
    total=sum(s['marks'])
    avg=total/len(s['marks'])
    grade=get_grade(avg)
    print(f"{s['roll']:<6}{s['name']:<18}{s['age']<5}"
          f"{total:<8.1f}{avg:<8.1f}{grade}")
def find_student(roll):
  for s in students:
    if s['roll']==roll:
      return s
  return None
def search_student():
  roll=int(input('Enter Roll Number to search:'))
  s=find_student(roll)
  if s:
    total=sum(s['marks'])
    avg=total/len(s['marks'])
    print(f"Name:{s['name']}")
    print(f"Age:{s['age']}")
    print(f"Marks:{s['marks']}")
    print(f"Total:{total} Avg:{avg:.1f} Grade:{get_grade(avg)}")
  else:
    print('Student not found.')
def update_student():
  roll=int(input('Enter Roll Number to update:'))
  s=find_student(roll)
  if not s:
    print('Student not found.')
    return
  print('1.Update Name 2.update Age 3.Update marks')
  choice=input('Choose dield:')
  if choice=='1':
    s['name']=input('New name:').strip()
  elif choice=='2':
    s['age']=int(input('New age:'))
  elif choice=='3':
    pass
  print('Record updated!')
def delete_student():
  roll=int(input('Enter Roll Number to delete:'))
  s=find_student(roll)
  if not s:
    print('Student not found.')
    return
  confirm=input(f"Delete {s['name']}?(yes/no):")
  if confirm.lower()=='yes':
    students.remove(s)
    print('Student deleted.')
  else:
    print('Seletion cancelled.')
def show_statistics():
  if not students:
    print('No data.')
    return
  avgs=[sum(s['marks'])/len(s['marks']) for s in students]
  best=students[avgs.index(max(avgs))]
  worst=students[avgs.index(min(avgs))]
  print(f'Total Students: {len(students)}')
  print(f'Class Average: {sum(avgs)/len(avgs):.2f}')
  print(f'Top Scorer :{best["name"]}({max(avgs):.1f})')
  print(f'Low Scorer :{worst["name"]}({min(avgs):.1f})')
def main_menu():
  menu={
      '1':add_student,
      '2':view_all_students,
      '3':search_student,
      '4':update_student,
      '5':delete_student,
      '6':show_statistics,
  }
  while True:
    print('\n===STUDENT MANAGMENT SYSTEM===')
    print('1.Add 2.View 3.Search 4.Update 5.Delete 6.Stats 7.Exit')
    choice=input('Enter choice:').strip()
    if choice=='7':
      print('Goodbye!')
      break
    elif choice in menu:
      menu[choice]()
    else:
      print('Invalid choice.Try again.')
if __name__=='__main__':
  main_menu()













```
