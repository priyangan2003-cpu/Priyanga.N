```python
#DATA INSERTION & BASIC RETRIEVAL
import sqlite3
conn=sqlite3.connect('college.db')
cursor=conn.cursor()
cursor.execute("""
CREATE TABLE students(
student_id INTEGER PRIMARY KEY,
name TEXT,
department TEXT,
marks INTEGER
)
""")
conn.commit()
print("Database and table ready.")
#INTER DATA INTO TABLE
students_data=[
    (1,'Arun','CSE',85),
    (2,'Divya','ESE',90),
    (3,'Karthik','IT',78),
    (4,'Meena','CSE',88),
    (5,'Rahul','EEE',72)
]
cursor.executemany(
    "INSERT INTO students VALUES(?,?,?,?)",
    students_data
)
conn.commit()
print("5 records inserted successfully.")
#SELECT*(View all records)
cursor.execute("SELECT * FROM students")
rows=cursor.fetchall()
print("All student Records:")
for row in rows:
  print(row)
#SELECT DISTINCT
cursor.execute("SELECT DISTINCT department FROM students")
rows=cursor.fetchall()
print("Distinct Department:")
for row in rows:
  print(row)
#SELECT TOP 3 records
cursor.execute("SELECT * FROM students LIMIT 3")
rows=cursor.fetchall()
print("Top 3 Records:")
for row in rows:
  print(row)
#ALIAS
cursor.execute("""
SELECT name AS Student_Name,
department AS dept,
marks AS Score
FROM students
""")
rows=cursor.fetchall()
print("Student Details with Alias:")
for row in rows:
  print(row)
```

```python
#FIL TERING & CONDITIONAL LOGIC
#Database & Table setup
import sqlite3
conn=sqlite3.connect('college.db')
cursor=conn.cursor()
cursor.execute("""
CREATE TABLE IF NOT EXISTS students(
student_id INTEGER PRIMARY KEY,
name TEXT,
depatment TEXT,
marks INTEGER
)
""")
students_data=[
    (1,'Arun','CSE',85),
    (2,'Divya','ECE',90),
    (3,'Karthik','IT',78),
    (4,'Meena','CSE',88),
    (5,'Rahul','EEE',72)
]
cursor.executemany(
    "INSERT OR IGNORE INTO students VALUES(?,?,?,?)",
    students_data
)
conn.commit()
print("Database and table ready with sample data.")
#WHERE
cursor.execute("SELECT * FROM students WHERE department='CSE'")
rows=cursor.fetchall()
print("Students from CSE depatment:")
for row in rows:
  print(row)
#AND
cursor.execute("""
SELECT * FROM students
WHERE department='CSE' AND marks > 85
""")
rows=cursor.fetchall()
print("CSE students with marks greater than 85:")
for row in rows:
  print(row)
#OR
cursor.execute("""
SELECT * FROM students
WHERE department ='CSE' OR department ='ECE'
""")
rows=cursor.fetchall()
print("Students from CSE or ECE:")
for row in rows:
  print(row)
#NOT
cursor.execute("""
SELECT * FROM students
WHERE NOT department='CSE'
""")
rows=cursor.fetchall()
print("Students NOT from CSE:")
for row in rows:
  print(row)
#IN
cursor.execute("""
SELECT * FROM students
WHERE department IN ('CSE','IT')
""")
rows=cursor.fetchall()
print("Students from CSE or IT:")
for row in rows:
  print(row)
#BETWEEN
cursor.execute("""
SELECT * FROM students
WHERE marks BETWEEN 80 AND 90
""")
rows=cursor.fetchall()
print("STUDENTS with mrks between 80 and 90:")
for row in rows:
  print(row)
#ORDER BY
cursor.execute("""
SELECT * FROM students
ORDER BY marks ASC
""")
rows=cursor.fetchall()
print("Students sorted by marks (Ascending):")
for row in rows:
  print(row)
#Desending Order
cursor.execute("""
SELECT * FROM students
ORDER BY marks DESC
""")
rows=cursor.fetchall()
print("Students sorted by marks(Descending):")
for row in rows:
  print(row)
```
