```python
#Creating single column DataFrame and checking lenght
import pandas as pd
names=["Priya","Raja","Ravi","Bala","Jeeva","Jothi","Pari","Vino"]
df=pd.DataFrame(names)
print(df)
print()
print("The length is:",len(df))
```

```python
#Creating DataFrame by using list of lists
import pandas as pd
details=[["Ranjan",11],["Sagar",12],["Daniel",13],["Prasd",14],["Kumar",15],["Priya",16],["Arjun",17],["Akshay",18]]
df=pd.DataFrame(details)
print(df)
```

```python
#Creating DataFrame and giving names to columns
import pandas as pd
details=[["Sagar",20,10000],["Daniel",16,20000],["Veeru",24,30000],["Raja",25,40000],["Kiran",26,50000],["Kedar",27,60000],["Reen",28,70000],["Karthik",29,80000],["Satish",30,90000]]
cols=["Name","Age","Salary"]
df=pd.DataFrame(details,columns=cols)
print(df)
```

```python
#Creating DataFrame by using dictionary
import pandas as pd
details={"Name":["Daniel","Abhi","Veeru","Raju","Kiran","Kedar","Reena","Karthik","Satish"],"Age":[20,21,23,24,25,26,27,28,29]}
df=pd.DataFrame(details)
print(df)
```

```python
#Creting DataFrame and giving index
import pandas as pd
details=[["Sagr",20,10000],["Daniel",16,20000],["Veeru",24,30000],["Raju",25,40000],["Kiran",26,50000],["Kedar",27,60000],["Reena",28,70000],["Karthik",29,80000],["Satish",30,90000]]
c=["Name","Age","Salary"]
i=[11,22,33,44,55,66,77,88,99]
df=pd.DataFrame(details,columns=c,index=i)
print(df)
```

```python
#Creating DataFrame and giving index
import pandas as pd
details=[["Sagar",20,10000],["Daniel",16,20000],["Veeru",24,30000],["Raja",25,40000],["Kiran",26,50000],["Reena",28,70000],["Karthik",29,80000],["Satish",30,90000]]
c=["Name","Age","Salary"]
i=["Row1","Row2","Row4","Row5","Row6","Row7","Row8","Row9"]
df=pd.DataFrame(details,columns=c,index=i)
print(df)
```
