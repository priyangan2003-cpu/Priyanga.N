```python
import pandas as pd
import sqlite3
#Loading the file excel
file_path="/content/SQLite_Joins_Practice_2_Sheets_100_Rows.xlsx"
excel=pd.ExcelFile(file_path)
print(excel.sheet_names)
customers_df=pd.read_excel(file_path,sheet_name="Customers")
orders_df=pd.read_excel(file_path,sheet_name="Orders")
print(customers_df.columns)
print()
print(orders_df.columns)
conn=sqlite3.connect("Supermarket.db")
cursor=conn.cursor()
print("Connection and database credted successfully")
customers_df.to_sql("Customers",conn,if_exists="replace",index=False)
orders_df.to_sql("Orders",conn,if_exists="replace",index=False)
print("Tables created successfully")
cursor.execute("SELECT*FROM Customers LIMIT5")
result=cursor.fetchall()
for r in result:
  print(r)
print()
cursor.execute("SELECT COUNT(*)FROM Orders")
result=cursor.fetchone()[0]
print(result)
#INNER JOIN
cursor.execute("""
SELECT*
FROM Customers AS c
INNER JOIN Orders AS o
ON c.Customer_ID=o.Customer_ID
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
#LEFT JOIN
print()
cursor.execute("""
SELECT*
FROM Customers AS c
LEFT JOIN Orders AS o
ON c.Customer_ID=o.Customer_ID
WHERE Order_ID IS NULL
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
#Right JOIN
print()
cursor.execute("""
SELECT*FROM Orders AS o
LEFT JOIN Customers AS c
ON o.Customer_ID=c.Customer_ID
WHERE o.Customer_ID IS NULL
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
#FULL OUTER JOIN
print()
cursor.execute("""
SELECT*
FROM Customers AS c
LEFT JOIN Orders AS o
ON c.Customer_ID=o.Customer_ID
UNION ALL
SELECT*
FROM Orders AS o
LEFT JOIN Customers AS c
ON o.Customer_ID=c.Customer_ID
WHERE o.Customer_ID IS NULL
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
#GROUP BY AND JOIN
print()
cursor.execute("""
SELECT c.Customer_ID,SUM(o.Amount) AS Total_amount
FROM Customers AS c
LEFT JOIN Orders AS o
ON c.Customer_ID=o.Customer_ID
GROUP BY c.Customer_ID
ORDER BY Total_amount DESC
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
#WHERE + JOIN
print()
cursor.execute("""
SELECT c.Customer_ID
FROM Customers AS c
LEFT JOIN Orders AS o
ON c.Customer_ID=o.Customer_ID
WHERE o.Order_Status="Completed" AND o.Amount>50000
LIMIT 5
""")
result=cursor.fetchall()
for r in result:
  print(r)
conn.close()
```
