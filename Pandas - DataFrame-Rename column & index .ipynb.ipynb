```python
#Creating a simple dataframe
import pandas as pd
d={
    "order_id":[11,21,31],
    "customer_name":["priya","Balaji","adhira"],
    "Product":["iPhone","htc","macbook"]
}
df1=pd.DataFrame(d)
print(df1)
```

```python
#Creating index in dataframe
import pandas as pd
d={
    "order_id":[11,21,31],
    "customer_name":["priya","Balaji","adhira"],
    "Product":["iPhone","htc","macbook"]
}
i={0:77,1:88,2:99}
df1=pd.DataFrame(d)
df2=df1.rename(index=i)
print(df1)
print()
print(df2)
```

```python
#Changing index in datframe using axis parameter
import pandas as pd
d={
    "order_id":[11,21,31],
    "customer_name":["priya","Balaji","adhira"],
    "Product":["iPhone","htc","macbook"]
}
df1=pd.DataFrame(d)
print(df1)
df1.index=[77,88,99]
print()
print(df1)
```

```python
#Changing column and index in a datframe
import pandas as pd
d={
    "order_id":[11,21,31],
    "customer_name":["priya","Balaji","adhira"],
    "Product":["iPhone","htc","macbook"]
}
df1=pd.DataFrame(d)
print(df1)
df1.index=[333,444,555]
df1.columns=["order_id","customer_name","product"]
print()
print(df1)
```
