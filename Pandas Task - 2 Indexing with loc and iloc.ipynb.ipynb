```python
#load data and inspect the defult index
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.index)
```

```python
#iloc-select a row by POSITION
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iloc[5])
```

```python
#set_index()-create a meaningful label index
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
df=df.set_index('Order ID')
print(df.index)
```

```python
#loc-select a row by LABEL
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc['ORD2005'])
```

```python
#loc vs iloc on the same row
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc['ORD2005'])
print(df.iloc[5])
```

```python
#iloc-select multiple rows by position(slicing)
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iloc[10:15])
```

```python
#loc-select multiple rows by label
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc['ORD2010':'ORD2014'])
```

```python
#loc-select specific columns by name
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc[:,['Product','Price']])
```

```python
#loc-rows AND  columns together
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc['ORD2000':'ORD2003',['Product','Region']])
```

```python
#iloc-select specific columns by position
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iloc[:,[1,4]])
```

```python
#iloc-rows AND columns together
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iloc[0:4,1:4])
```

```python
#Conditional Filtering with loc+column selection
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc[df['Region']=='West',['Product','Quantity']])
```

```python
#conditional (boolean)filtering with loc
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.loc[df['Quantity']>15])
```

```python
#at-get a single scalar value by label
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.at['ORD2020','Price'])
```

```python
#iat-get a single scalar value by position
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iat[3,2])
```

```python
#Modify values using loc
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
df.loc['ORD2001','Region']='Central'
print(df.loc['ORD2001'])
```

```python
#Modify values using iloc
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
df.iloc[0,df.columns.get_loc('Quantity')]=100
print(df.iloc[0])
```

```python
#iloc with negative
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
print(df.iloc[-5:])
```

```python
#reset_index()
import pandas as pd
df=pd.read_excel('orders_data.xlsx.xlsx')
df=df.reset_index()
print(df.index)
print(df.columns)
```
