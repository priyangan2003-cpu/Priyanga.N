```python
#Coverting float column type into int data type
import pandas as pd
df1=pd.read_csv('fruits1.csv')
df2=df1.dropna()
df3=df2.astype(int)
print(df2.head())
print()
print(df3.head())
```

```python
#Dropping Nn values by using inplace parameter
import pandas as pd
df1=pd.read_csv("fruits1.csv")
df1.dropna(inplace=True)
print(df1)
```

```python
#Filling NaN values with zero
import pandas as pd
df1=pd.read_csv("fruits1.csv")
df2=df1.fillna(0)
print(df1.head())
print()
print(df2.head())
```

```python
#Filling NaN values with specific value
import pandas as pd
import numpy as np
data=[["Rajan",26,40000],["Daniel",16,20000],["Veeru",45,90000],["Venkat",np.nan,45000],["Sumanth",20,95000],["Shafi",np.nan,97000]]
df1=pd.DataFrame(data,columns=['Name','Age','Salary'])
df2=df1.fillna(22)
print(df1)
print()
print(df2)
```

```python
#Filling NaN value with mean value
import pandas as pd
import numpy as np
data=[["Priya",23,50000],["balu",27,50000],["Saravana",20,40000],["Reena",23,50000],["Suresh",np.nan,97000],["Ravi",24,50000]]
df1=pd.DataFrame(data,columns=['Name','Age','Salary'])
print(df1)
m=df1['Age'].mean()
df1['Age']=df1['Age'].fillna(m)
print()
print(df1)
```

```python
#Creating dataframe and replacing nan values with specific value
import pandas as pd
import numpy as np
data=[['Priya',np.nan,40000],['Balu',16,20000],['Veeru',45,90000],['Sumanth',20,950000]]
df1=pd.DataFrame(data,columns=['Name','Age','Salary'])
print(df1)
df2=df1.replace(np.nan,0)
print()
print(df2)
```
