```python
#
import pandas as pd
df = pd.read_excel('sales_data.xlsx')
region_revenue = df.groupby('Region')['Revenue'].sum()
print(region_revenue)
print()
avg_qty = df.groupby('Category')['Quantity'].mean()
print(avg_qty)
print()
combo_revenue = df.groupby(['Region', 'Category'])['Revenue'].sum()
print(combo_revenue)
print()
summary = df.groupby('Region')['Revenue'].agg(['sum', 'mean', 'max'])
print(summary)
print()
summary = df.groupby('Category').agg({'Revenue': 'sum', 'Quantity': 'mean'})
print(summary)
print()
print(df.groupby('Sales Rep').size())
print(df.groupby('Sales Rep')['Revenue'].count())
print()
region_totals = df.groupby('Region')['Revenue'].sum().reset_index()
print(region_totals)
print()
sorted_asc = df.sort_values('Revenue')
print(sorted_asc.head())
print()
sorted_desc = df.sort_values('Quantity', ascending=False)
print(sorted_desc.head())
print()
sorted_multi = df.sort_values(by=['Region', 'Revenue'],
                               ascending=[True, False])
print(sorted_multi.head())
print()
cat_totals = df.groupby('Category')['Revenue'].sum()
print(cat_totals.sort_index())
print()
region_totals = df.groupby('Region')['Revenue'].sum()
top_regions = region_totals.sort_values(ascending=False)
print(top_regions)
print()
region_totals = df.groupby('Region')['Revenue'].sum()
print(region_totals.idxmax())
print()
df['Region Avg Revenue'] = df.groupby('Region')['Revenue'].transform('mean')
print(df[['Region','Revenue','Region Avg Revenue']].head(10))
print()
print(df['Sales Rep'].value_counts())
print()
pivot = df.pivot_table(values='Revenue', index='Region',
                        columns='Category', aggfunc='sum')
print(pivot)
print()
grouped = df.groupby('Category')
electronics_only = grouped.get_group('Electronics')
print(electronics_only.head())
print()
df_sorted = df.sort_values(['Region', 'Sale Date'])
df_sorted['Cumulative Revenue'] = df_sorted.groupby('Region')['Revenue'].cumsum()
print(df_sorted[['Region','Sale Date','Revenue','Cumulative Revenue']].head(10))
print()
top_per_region = (df.sort_values('Revenue', ascending=False)
                    .groupby('Region')
                    .head(1))
print(top_per_region)
```
